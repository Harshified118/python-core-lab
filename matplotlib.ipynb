```python
import numpy as np 
```

```python
def f(t):return np.exp(-t)*np.cos(2*np.pi*t)
t1=np.arange(0.0,5.0,0.1)
t2=np.arange(0.0,5.0,0.02)
plt.figure(1)
plt.subplot(211)
plt.grid()
plt.plot(t1,f(t1),'b-')

plt.subplot(212)
plt.plot(t2, np.cos(2*np.pi*t2),'r--')
plt.show()
```

```python
import matplotlib.pyplot as plt
category=['Apple','Banana','Mango','Jerry']
value=[20,30,10,40]
plt.bar(category,value,color='red', edgecolor='black')
plt.xlabel('Fruits Type')
plt.ylabel('Fruits Quantity')
plt.title('Bar Plt of Session Fruits')
plt.show()
```

```python
category=['Apple', 'Banana', 'Mango', 'Jerry']
value=[20,30,10,40]
plt.pie(value, labels=category)
plt.title('Session Fruits')
plt.show()
```

```python
category=['Apple','Banana','Mango','Jerry']
value=[20,30,10,40]
myexplode=[0.1,0,0,0]
plt.pie(value,labels = category, explode=myexplode, shadow=True)
plt.title('Session Fruits')
plt.show()
```
